```python
import pandas as pd
import json
```

```python
with open("./company-tickers.json") as file:
    list_ticker = json.load(file)
```

```python
# Parse ticker (company code) and cik_str (company ID)
companies = [(entry["cik_str"], entry["ticker"]) for entry in list_ticker.values()]
```

```python
company_df = pd.DataFrame(companies, columns = ['company_id', 'company_ticker'])
```

```python
seen = set()
with open("tickers.txt", "w") as f:
    for cik, ticker in companies:
        cik_str = str(cik).zfill(10)
        if cik_str not in seen:
            seen.add(cik_str)
            f.write(cik_str + "\n")
```
